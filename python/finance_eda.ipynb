# Finance exploratory analysis
Reproducible exploratory analysis using the provided synthetic dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
path = Path('../data/financial_transactions.csv')
df = pd.read_csv(path, parse_dates=['month'])
assert df.isna().sum().sum() == 0
assert (df[['revenue','expense']] >= 0).all().all()
monthly = df.groupby('month', as_index=False)[['revenue','expense','revenue_budget','expense_budget']].sum()
monthly['profit'] = monthly.revenue - monthly.expense
monthly['margin'] = monthly.profit / monthly.revenue
print(monthly.to_string(index=False))
print('Total operating profit:', round(monthly.profit.sum(),2))
print('Revenue variance:', round((monthly.revenue-monthly.revenue_budget).sum(),2))
monthly.plot(x='month', y=['revenue','expense','profit'], figsize=(11,5), title='Monthly financial performance')
plt.tight_layout(); plt.show()
```
